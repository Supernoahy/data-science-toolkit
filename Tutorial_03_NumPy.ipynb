```python
import numpy as np
```

```python
#2*2 NumPy array (fixed values)
array_2x2 =np.array([[1,2],[3,4]])
print("Task 1-2*2 Array:")
print(array_2x2)
print("shape:",array_2x2.shape)
```

```python
# Random 3x4 array (integers between 0 and 50)
random_3x4 = np.random.randint(low=0, high=51, size=(3, 4))  # size=(rows, columns)
print("\nTask 2 - Random 3x4 Array:")
print(random_3x4)
print("Shape:", random_3x4.shape)
```

```python
# Array of ones (3x3)
ones_array = np.ones(shape=(3, 3), dtype=int)  # dtype=int to avoid floats
# Array of zeros (2x5)
zeros_array = np.zeros(shape=(2, 5), dtype=int)

print("\nTask 3 - Array of Ones (3x3):")
print(ones_array)
print("\nTask 3 - Array of Zeros (2x5):")
print(zeros_array)
```

```python
# Random 6x3 array (floats between 0 and 1)
random_6x3 = np.random.rand(6, 3)
# Transpose the array
transposed_array = random_6x3.T  # Equivalent to np.transpose(random_6x3)

print("\nTask 4 - Original Random 6x3 Array:")
print(random_6x3)
print("Shape (original):", random_6x3.shape)
print("\nTask 4 - Transposed Array (3x6):")
print(transposed_array)
print("Shape (transposed):", transposed_array.shape)
```

```python
# Define dimensions (n, p, m)
n = 2
p = 3
m = 4

# Create two random arrays (n×p and p×m)
array_np = np.random.rand(n, p)  # 2x3 array
array_pm = np.random.rand(p, m)  # 3x4 array

# Compute dot product (use np.dot() or @ operator)
dot_product = np.dot(array_np, array_pm)  # Equivalent to array_np @ array_pm

print("\nTask 5 - First Array (n×p = 2×3):")
print(array_np)
print("\nTask 5 - Second Array (p×m = 3×4):")
print(array_pm)
print("\nTask 5 - Dot Product (n×m = 2×4):")
print(dot_product)
print("Shape (dot product):", dot_product.shape)  # Should be (2, 4)
```
